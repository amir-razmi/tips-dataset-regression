```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme(
  style='darkgrid',
  palette='viridis'
)
```

```python
df = sns.load_dataset('tips')
df
```

```python
ax = sns.histplot(data=df, x='total_bill', kde=True, element='step')
ylim = ax.get_ylim()[1]
ax.vlines(x=df['total_bill'].mean(), ymin=0, ymax=ylim, colors='red', label='Mean')
ax.vlines(x=df['total_bill'].median(), ymin=0, ymax=ylim, colors='green', label='Median')
ax.vlines(x=df['total_bill'].mode(), ymin=0, ymax=ylim, colors='blue', label='Mode')

ax.legend()

# Total_bill distribution is positively skewed
```
