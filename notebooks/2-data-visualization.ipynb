```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme(
  style='darkgrid',
  palette='viridis'
)
```

```python
df = sns.load_dataset('tips')
df
```

```python
ax = sns.histplot(data=df, x='total_bill', kde=True, element='step')
ylim = ax.get_ylim()[1]
ax.vlines(x=df['total_bill'].mean(), ymin=0, ymax=ylim, colors='red', label='Mean')
ax.vlines(x=df['total_bill'].median(), ymin=0, ymax=ylim, colors='green', label='Median')
ax.vlines(x=df['total_bill'].mode(), ymin=0, ymax=ylim, colors='blue', label='Mode')

ax.legend()

#INFO: Total_bill distribution is positively skewed
```

```python
ax = sns.histplot(data=df, x='tip', element='step', kde=True)
ylim = ax.get_ylim()[1]
ax.vlines(x=df['tip'].mean(), ymin=0, ymax=ylim, color='red', label='Mean')
ax.vlines(x=df['tip'].median(), ymin=0, ymax=ylim, color='green', label='Median')
ax.vlines(x=df['tip'].mode(), ymin=0, ymax=ylim, color='blue', label='Mode')

ax.legend()

#INFO: tip distribution is positively skewed
```

```python
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(12, 10))

sns.countplot(data=df, x='sex', ax=ax1)
sns.countplot(data=df, x='smoker', ax=ax2)
sns.countplot(data=df, x='day', ax=ax3)
sns.countplot(data=df, x='time', ax=ax4)

#INFO: Male > Female
#INFO: None-smoker > Smoker
#INFO: Saturday > Sunday > Thursday > Friday
#INFO: Dinner > Lunch
```

```python
ax = sns.countplot(data=df, x='size')

ax.bar_label(ax.containers[0]); # Automatically labels the bars with their respective counts

#INFO: 2 > 3 > 4 > 5 > 1 = 6
```
